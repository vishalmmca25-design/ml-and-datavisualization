```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Load dataset
df = pd.read_csv(
    "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
)

# Display information
df.info()

# 2. Count missing values
print("\nMissing values:")
print(df.isnull().sum())

# 3. Percentage of Cabin missing
cabin_missing = df["Cabin"].isnull().mean() * 100
print("\nCabin missing percentage:", cabin_missing)

# Drop Cabin
df = df.drop("Cabin", axis=1)

# 4. Fill Age with median
df["Age"] = df["Age"].fillna(df["Age"].median())

# Remove remaining missing rows
df = df.dropna()

print("\nRows remaining:", df.shape[0])

# 5. Confirm no missing values
print("\nTotal missing values:", df.isnull().sum().sum())

# 6. Create FamilySize
df["FamilySize"] = df["SibSp"] + df["Parch"] + 1

# 7. Overall survival rate
print("\nOverall survival rate:")
print(df["Survived"].mean())

# 8. Survival by Sex
print("\nSurvival rate by Sex:")
print(df.groupby("Sex")["Survived"].mean())

# Survival by Pclass
print("\nSurvival rate by Pclass:")
print(df.groupby("Pclass")["Survived"].mean())

# 9. Survival by Sex and Pclass
print("\nSurvival rate by Sex and Pclass:")
print(df.groupby(["Sex", "Pclass"])["Survived"].mean())

# 10. Bar chart
survival_by_class = df.groupby("Pclass")["Survived"].mean()

survival_by_class.plot(kind="bar")

plt.title("Survival Rate by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Survival Rate")
plt.xticks(rotation=0)
plt.show()
```
