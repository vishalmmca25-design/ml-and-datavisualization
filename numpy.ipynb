1.A small shop recorded the number of items it sold on each of the seven days of a week. The owner wants to understand how business went during the week.
	The data:
sales = np.array([23, 17, 35, 12, 40, 28, 31])

Using NumPy, find out:
(a)  the total number of items sold during the whole week;
(b)  the average number of items sold per day;
(c)  on how many days the shop sold more than the daily average, and what those above-average sales figures were.
# solution

```python

import numpy as np

# Sales data
sales = np.array([23, 17, 35, 12, 40, 28, 31])

# (a) Total items sold
total_sales = np.sum(sales)

# (b) Average items sold per day
average_sales = np.mean(sales)

# (c) Days with above-average sales
above_average = sales[sales > average_sales]
count_above_average = np.sum(sales > average_sales)

#how many days the shop sold more than the daily average
daymore=sales[sales>average_sales]
print(daymore.size)

# Display results
print("Total items sold:", total_sales)
print("Average items sold per day:", average_sales)
print("Number of days with above-average sales:", count_above_average)
print("Above-average sales figures:", above_average)




```

2.Background.  You are given the raw marks of a class as a 1-D array and must convert them to letter grades and summarise the class. Doing this with an if-else inside a loop is easy but slow; the point of this exercise is to express the whole mapping with boolean masks so it runs in one shot over the array.
                  Given input:
               marks = np.array([92, 61, 74, 45, 88, 90, 59, 77, 63, 100, 70, 55, 84, 66, 39, 75,
                  81, 60, 49, 95])
 (a)  Produce an array grades of the same length holding a single letter per student, using these bands: A for marks ≥ 90, B for 75–89, C for 60–74, and F for anything below 60. Build the result with boolean-mask assignment (or nested np.where), not a loop.
(b)  Using NumPy, report how many students fall in each of the four bands.
(c)  Using a boolean condition, print the marks of every student who passed (mark ≥ 60), in descending order.
Constraints:  No explicit loops. Watch the band boundaries — they must not overlap.
#solution


```python



#(a) Produce an array grades of the same length holding a single letter per student,
#using these bands: A for marks ≥ 90, B for 75–89, C for 60–74, and F for anything below 60.
#Build the result with boolean-mask assignment (or nested np.where), not a loop



marks = np.array([92, 61, 74, 45, 88, 90, 59, 77, 63, 100,
                  70, 55, 84, 66, 39, 75, 81, 60, 49, 95])

#  empty numpy array to store grades
grades = np.empty(len(marks), dtype='<U1')

# Assign grades using boolean masks
grades[marks >= 90] = 'A'
grades[(marks >= 75) & (marks <= 89)] = 'B'
grades[(marks >= 60) & (marks <= 74)] = 'C'
grades[marks < 60] = 'F'

print("Marks :", marks)
print("Grades:", grades)

#(b)  Using NumPy, report how many students fall in each of the four bands.
count_A = np.sum(marks >= 90)
count_B = np.sum((marks >= 75) & (marks <= 89))
count_C = np.sum((marks >= 60) & (marks <= 74))
count_F = np.sum(marks < 60)

print("A:", count_A)
print("B:", count_B)
print("C:", count_C)
print("F:", count_F)

#(c)  Using a boolean condition, print the marks of every student who passed (mark ≥ 60), in descending order.
#Constraints:  No explicit loops. Watch the band boundaries — they must not overlap.
# Select passed students
passed = marks[marks >= 60]

# Sort in descending order
passed_desc = np.sort(passed)[::-1]

print("Passed marks (descending):")
print(passed_desc)



```

3. A teacher has the marks of eight students in three subjects, stored as a matrix in which each row is one student and each column is one subject. She needs a quick analysis of the results.
The data (rows = students 0–7, columns = subjects):
scores = np.array([[45, 50, 40],
                   [70, 65, 80],
                   [30, 55, 60],
                   [90, 85, 88],
                   [40, 42, 38],
                   [55, 60, 58],
                   [75, 70, 72],
                   [20, 35, 50]])

Using NumPy (no loops), determine:
(a)  each student's total marks and average across the three subjects;
(b)  which student scored the highest total — report the student's row index and their total;
(c)  which subject had the highest class average (report the column index);
(d)  how many students passed every subject, where a pass in a subject means a mark of 40 or more.

#solution


```python
scores = np.array([[45, 50, 40],
                   [70, 65, 80],
                   [30, 55, 60],
                   [90, 85, 88],
                   [40, 42, 38],
                   [55, 60, 58],
                   [75, 70, 72],
                   [20, 35, 50]])

#(a) each student's total marks and average across the three subjects
total_marks = np.sum(scores, axis=1)
average_marks = np.mean(scores, axis=1)

print("Total marks:", total_marks)
print("Average marks:", average_marks)

#(b)  which student scored the highest total — report the student's row index and their total;
highest_total = np.argmax(total_marks)
print("Student with highest total:")
print("Row index:", highest_total)
print("Total marks:", total_marks[highest_total])

#(c)  which subject had the highest class average (report the column index);
highest_average = np.argmax(np.mean(scores, axis=0))
print("Subject with highest class average:")
print("Column index:", highest_average)


# (d) Students who passed every subject (>= 40)
passed_every_subject = np.sum(np.all(scores >= 40, axis=1))

print("\nStudents Passed Every Subject:", passed_every_subject)

```
